### Diskussion

Die gemessenen Kreisfrequnzen $\omega _s = (0.0953\pm 0.0029)\frac{1}{s} $ und $\omega _2 = (3.62\pm 0.42)\frac{1}{s} $.
die Unsicherheit kommt hauptsächlich von der Stoppuhr. Systematische Fehler könnten eine schräge Auslenkung und die menschliche Reaktionszeit, da alle Wete händisch aufgenommen wurden, sein.

Die durch die Schwebung berechneten Eigenfrequenzen $\omega _0 = (3.53\pm 0.42)\frac{1}{s}$ und $\omega _1 = (3.72\pm 0.42)\frac{1}{s}$ entsprechen den zusätzlich bestimmten Frequenzen $\omega _{gl}$ und $\omega _{geg}$ .
