
### 1. Erklärung der Grundlagen

Wie auch in PW8 geht es in diesen Versuchsaufbauten wieder um Wechselspannungen und Reihenschaltungen. Dieses mal sollen ebenfalls, wie bereits in PW8, verschiedene elektrische Größen wie $R$, $X_C$, $X_L$ sowie $C$, $L$ und die Phasenverschiebungen $\Delta \varphi$ berechnet werden.

Für die drei verschiedenen Arten von Wechselstromwiderständen gilt weiterhin:
$$
\text{Ohm'scher Widerstand: $R$ = } \ R
$$
$$
\text{Kondensator: } X_C = \frac{1}{\omega \cdot C}
$$
$$
\text{Spule: } X_L = \omega \cdot L
$$

<br>
<br>

In diesem Aufbau wird, wie im Anhang zu sehen ist, eine Serienschaltung, in dem stets ein bekannter Widerstand R_0 vorhanden ist, mit zunächst einem Ohm'schen Widerstand, dann einem Kondensator und zum Schluss einer Spule aufgebaut. Durch das Messen der Gesamtspannung $U_1$ und der am bekannten Ohm'schen Widerstand abfallenden Spannung $U_2$ lässt sich mittels folgender Gleichung der Betrag der komplexen Impdedanz |X|, also $X_C$ oder $X_L$, berechnen:
$$
|X| = \ R_0 \cdot \sqrt{(\frac{U_1}{U_2})^2 - 1}
$$

<br>

Für einen Ohm'schen Widerstand gilt lediglich: 
$$
R_x = R_0 \cdot (\frac{U_1}{U_2} - 1)
$$

<br> 

Da es sich um eine Wechselspannung handelt, gibt es eine Phasenverschiebung $\Delta \varphi$, welche im Experiment gemessen wird. Zusätzlich wird eben diese Phasenverschiebung $\Delta \varphi$ mit folgender Gleichung berechnet:
$$
\Delta \varphi = \arctan{\frac{X}{R_0}}
$$


```python
R_0 = 500 # Ohm
v = 500 # Hz

U_0 = 500 # V (Gesammtspannung)
U_1 = 200 # V (an R_0)

# 2.1

R_x = ufloat(R_0 * ((U_0/(U_1)) - 1),R_0 * ((U_0/(U_1)) - 1) * 0.05 ) # Gesuchter Widerstand (Ohm)
dp = 0 # Phasenverschiebung




display(Latex(fr"Die Phassenverschiebung beträgt: {dp}° und {dp}rad, da es sich um einen Ohm´schen Widerstand handelt."))

display(Latex(fr"Der gesuchte Widerstand beträgt: ({R_x}) Ohm."))
```

```python
R_0 = 500 # Ohm
v = 150 # Hz

U_0 = 500 # V (Gesammtspannung)
U_1 = 200 # V (an R_0)

# 2.2

R_x = ufloat(R_0 * ((U_0/(U_1))**2 - 1)**(1/2),R_0 * ((U_0/(U_1))**2 - 1)**(1/2) * 0.05 ) # Gesuchter Widerstand (Ohm)
X = -R_0 * np.sqrt((U_0/U_1)**2 - 1)
dp = ufloat(unp.arctan(X/R_0), -unp.arctan(X/R_0)* 0.05) # Phasenverschiebung °
dp_r = radians(dp) # Phasenverschiebung rad
C = ufloat(1/(X * v),1/-(X * v)*0.05) # Kapazität F


display(Latex(fr"Die berechnete Phassenverschiebung beträgt: ({dp})° und ({dp_r})rad, da es sich um einen Ohm´schen Widerstand handelt."))

display(Latex(fr"Der gesuchte Widerstand beträgt: ({R_x}) Ohm."))

display(Latex(fr"Die gesuchte Kapazität beträgt: ({C}) F."))
```

```python
R_0 = 500 # Ohm
v = 30000 # Hz

U_0 = 500 # V (Gesammtspannung)
U_1 = 200 # V (an R_0)

# 2.3

R_x = ufloat(R_0 * ((U_0/(U_1))**2 - 1)**(1/2),R_0 * ((U_0/(U_1))**2 - 1)**(1/2) * 0.05 ) # Gesuchter Widerstand (Ohm)
X = R_0 * np.sqrt((U_0/U_1)**2 - 1)
dp = ufloat(unp.arctan(X/R_0), unp.arctan(X/R_0)* 0.05) # Phasenverschiebung °
dp_r = radians(dp) # Phasenverschiebung rad
L = ufloat(X/(v),X/(v)*0.05) # Kapazität F


display(Latex(fr"Die berechnete Phassenverschiebung beträgt: ({dp})° und ({dp_r})rad, da es sich um einen Ohm´schen Widerstand handelt."))

display(Latex(fr"Der gesuchte Widerstand beträgt: ({R_x}) Ohm."))

display(Latex(fr"Die gesuchte Induktivität beträgt: ({L}) H."))
```

### Diskussion

Der ohmsche Widerstand beträgt $R_x = (5.86 \pm 0.29)k\Omega$. Das entspricht dem abgelesenen Wert von $6k\Omega$. 

Der kapazitiver Widerstand hat eine Kapazität $C = (0.22 \pm 0.01)\mu F$. Das entspricht auch dem gemessenen Wert von $0.23\mu F$. 

Der induktive Widerstand hat eine Induktivität $L = (11.59 \pm 0.58)mH$. Das entspricht in einer 10% bereich der gegebenen Induktivität $10mH$. Diese erhöhte Abweichung kann der Abnutzung der Spule zugeordnet werden. Außerdem konnte die Induktivität nicht direkt gemessen werden, sondern wurde als Wert vorgegeben.
