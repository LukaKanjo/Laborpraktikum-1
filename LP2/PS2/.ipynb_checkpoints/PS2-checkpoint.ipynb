```python
import numpy as np 
from uncertainties import ufloat
from uncertainties import unumpy as unp
from IPython.display import Latex, HTML, Math, display
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.stats import linregress
import pandas as pd
from scipy import constants as c
from matplotlib.patches import Rectangle
from matplotlib.ticker import MaxNLocator
from sklearn.metrics import r2_score
```

## Flavor


```python
# 2.2 Druck und Volumen konstante Temperatur

# Konstanten
T_0 = 22+273.15       # K
p_0 = 1002   # mbar
h_hg = 1.2        # cm

# Werte
dp = 1002 - np.array([1000 ,577, 472, 399, 351, 326, 304, 312])
h = np.array([8 ,13.5, 16.8, 19.4, 21.8, 23.6, 25.1, 24.6])

# Unsicherheiten
T_u = ufloat(T_0, 3)
p_u = ufloat(1002, 1)
dp_u = unp.uarray(dp, 1)
h_u = unp.uarray(h, 0.1)
h_hg_u = ufloat(1, 0.1)


# Formel Druck
def Druck(dp, h, p_0):
    p_hg = c.g*h*136/100
    p = p_0 - dp + p_hg
    return p

# Formel Volumen
def Volumen(h):
    V = c.pi*0.27**2*1/4*h
    return V

# Fit
def model_f(x,a):
    return a/x

# Rechnung ohne Unsicherheiten für den Fit
p = Druck(dp, h_hg, p_0)
V = Volumen(h)


#Unsicherheitsrechnung

p_u = Druck(dp_u, h_hg_u, p_0)
V_u = Volumen(h_u)
print(p_u)
print(V_u)
y_err = 20

# Fit mit y_err
popt, pcov = curve_fit(model_f, V, p, sigma=y_err, absolute_sigma=True, p0=[1])


# A mit Unsicherheit aus dem Fit
a_opt = popt
x_model = np.linspace(min(V), max(V), 100)
y_model = model_f(x_model, a_opt)
perr = np.sqrt(np.diag(pcov))
print(perr)
print(a_opt)



# r^2 Rechnung
def r2_curve_fit(popt, x, y):
    y_pred = model_f(x, *popt)
    r2 = r2_score(y, y_pred)
    return r2

# Zusammenfügen
r2 = r2_curve_fit(popt, V, p)
A = ufloat(a_opt,perr)
A_u = A/10000
print(f"A = {A_u}")
R =  8.3143/1000
n = A_u/(R*T_0)
print(f"n = {n:.6f}")
V_m = V_u[0]/n
print(f"{V_m:.3f}")


# Plot
plt.figure(figsize=(10, 6))
plt.scatter(V, p)
plt.plot(x_model, y_model, color='r', label=f"$y(x)=A/x$ \n A=({A:.2f}) $mbar$ $cm^3$ \n $r^2$={r2:.4f}")
# plt.plot(V, p, "o")
plt.ylabel(r"p in mbar")
plt.xlabel(r"V in cm³")
plt.legend()
plt.title("pV-Diagramm")
plt.tight_layout()
plt.grid()
plt.show()
print(r2)
```

```python
# 2.2 Temperatur und Volumen konstante Druck


p_0 = ufloat(1002, 1)   # mbar
Alp_e = 1/273.15

T = np.array([44, 46, 37, 31, 27, 24, 20, 17, 14, 8])      #°C
h = np.array([9.0, 9.1, 8.9, 8.8, 8.7, 8.6, 8.5, 8.3, 8.3, 8.1])   #cm

# Unsicherheiten
Tu = unp.uarray(T,3)
hu = unp.uarray(h,0.1)
print(f"T = {Tu}")
print(f"h = {hu}")
h__u = 0.1
V__u = Volumen(h__u)
print(V__u)

# r^2
def r2_curve_fit(func, popt, x, y):
    y_pred = func(x, *popt)
    r2 = r2_score(y, y_pred)
    return r2

# Formel Volumen
def Volumen(h):
    V = c.pi*0.27**2*1/4*h
    return V
V = Volumen(h)
print(V)


# Formel Fit
def linfit(x, a, b):
    return a*x+b

# Curve fit
popt, pcov = curve_fit(linfit, T, V,sigma=V__u,absolute_sigma=True, p0=[1,0])
a_opt, b_opt = popt
x_model = np.linspace(min(T), max(T), 100)
y_model = linfit(x_model, a_opt, b_opt)


#Unsicherheitsrechnung
perr = np.sqrt(np.diag(pcov))
print(perr)
a_u = ufloat(a_opt,perr[0])
b_u = ufloat(b_opt, perr[1])
alpha = a_u/b_u
print(alpha)
print(f"expected: {Alp_e}")

# r^2
r2 = r2_curve_fit(linfit, popt, T, V)

# plot
plt.figure(figsize=(10, 6))
plt.plot(T, V, "o")
plt.plot(x_model, y_model, color='r', label=f" $y(x)=ax+b$ \n a=({a_u})$cm^3/°C$ \n b=({b_u})$cm^3$ \n $r^2$={r2:.3f}")
plt.ylabel(r"V in cm³")
plt.xlabel(r"T in °C")
plt.legend()
plt.title("VT-Diagramm")
plt.tight_layout()
plt.grid()
plt.show()
print(Tu)
print(a_opt, b_opt)
lol = alpha/Alp_e
print(lol)
```
